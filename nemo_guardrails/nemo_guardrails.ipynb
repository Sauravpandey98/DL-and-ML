## NemoGuardrails
Here we will test NemoGuardrail functionality and how does different rails work.The four type of rails we will test are:

* Input Rails
* Dialog Rails
* Retriever Rails
* Output Rails

Also we will test `Action` feature of NemoGuardrails.

### Step1: Setup the configuration

Configuration can be setup in `config.yaml` file.In this file we can enable/disable the rails we want to activate, can setup the llm configuration,can setup system prompt, can setup sample conversations etc.

The followed folder structure is this:

- config/
  - rails/
    - file_1.co
    - file_2.co
    - ...
  - actions.py
  - config.py
  - config.yml




You can read more about the files from here: https://docs.nvidia.com/nemo/guardrails/latest/user-guides/configuration-guide.html

### Step2: Setup LLM and create Rails Instance 

```python
import os
from nemoguardrails import LLMRails, RailsConfig
from langchain.chat_models import openai
from dotenv import load_dotenv

load_dotenv(".env")

# Define LLM and parameters to pass to the guardrails configuration
chat_model = openai.ChatOpenAI(model="gpt-3.5-turbo",api_key=os.getenv("OPENAI_API_KEY"))

# Load configuration
config = RailsConfig.from_path("./config")

# Configuration of LLMs is passed
app = LLMRails(config=config, llm=chat_model)
```

```python
new_message = await app.generate_async(messages=[{
    "role": "user",
    "content": "What's the latest fashion trend?"
},{"role":"assistant","content":"I cannot answer to your question because I'm programmed to assist only with insurance-related questions."},
{"role":"user","content":"what are your capabilities?"}])

new_message
```
